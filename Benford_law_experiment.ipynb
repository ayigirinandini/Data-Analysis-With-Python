```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Parameters
n = 10000
scale = 0.1

# Generate Data
data = np.random.exponential(size=n)*scale
# Get integer part of digits

```

```python
data = data.round(4)*10000
```

```python
digs = [int(str(abs(x)).split('.')[0][0]) for x in data]

# Calculate frequency
counts = pd.Series(digs).value_counts().sort_index()

percentages = pd.Series(digs).value_counts(normalize=True).sort_index() * 100

# Plot as a bar chart
percentages.plot(kind='bar')

plt.title('Percentage Distribution of First Digits')
plt.xlabel('Digit')
plt.ylabel('Percentage (%)')
plt.show()
```
