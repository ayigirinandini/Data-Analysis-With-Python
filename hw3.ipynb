```python
import numpy as np
import math
from mpmath import exp,nsum,inf,factorial

mean = 100
def lost_demand(q):
   val = nsum(lambda d: (d-q)* exp(-mean) * mean**d/factorial(d),[q,inf])
   return val
print(lost_demand(90))
print(lost_demand(95))
print(lost_demand(98))
print(lost_demand(99))
```
