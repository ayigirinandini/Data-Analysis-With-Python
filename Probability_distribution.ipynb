```python
#checking the three sigma rule also called 68-95-97 rule of normal distribution
import numpy as np
mean = 0
std_dev = 1
sample = np.random.normal(mean,std_dev,10000)
#calculating values within one standard deviation
within_1_std = np.sum(np.abs(sample-mean) < std_dev)
#calculating values within one standard deviation
within_2_std = np.sum(np.abs(sample-mean) < 2*std_dev)
#calculating values within one standard deviation
within_3_std = np.sum(np.abs(sample - mean) < 3 * std_dev)
print(within_1_std/100)
print(within_2_std/100)
print(within_3_std/100)
```
